```python
from neo4j import GraphDatabase
import csv

BATCH_SIZE = 1000

driver = GraphDatabase.driver(
    neo4j_uri,
    auth=(neo4j_user, neo4j_password)
)

def insert_batch(tx, rows):

    tx.run("""
    UNWIND $rows AS row

    CREATE (:ChemicalTaxonomies {

        data_name: row.Data_Name,
        data_type: row.Data_Type,
        data_version: row.Data_Version,
        record_id: row.Record_ID,

        chemical_name: row.Chemical_Name,
        casrn: row.CASRN,
        dtxsid: row.DTXSID,

        inchikey: row.InChiKey,
        smiles: row.SMILES,


        kingdom: row.Kingdom,
        superclass: row.Superclass,
        class: row.Class,
        subclass: row.Subclass,

        level_5: row.Level_5,
        level_6: row.Level_6,
        level_7: row.Level_7,
        level_8: row.Level_8,
        level_9: row.Level_9,
        level_10: row.Level_10,


        structure_used: row.Structure_Used,
        tool_used: row.Tool_Used,

        date_extracted: row.Date_Extracted,

        reference: row.Reference,
        url: row.URL
    })

    """,
    rows=rows
    )

def load():
    with driver.session() as session:
        batch = []

        with open(
            TSV_FILE,
            "r",
            encoding="utf-8"
        ) as f:
            reader = csv.DictReader(
                f,
                delimiter="\t"
            )

            for row in reader:
                batch.append(row)

                if len(batch) == BATCH_SIZE:
                    session.execute_write(
                        insert_batch,
                        batch
                    )

                    print(
                        f"Inserted {len(batch)}"
                    )
                    batch.clear()

            if batch:
                session.execute_write(
                    insert_batch,
                    batch
                )
                print(
                    f"Inserted final {len(batch)}"
                )

load()

driver.close()


print("ChemicalTaxonomies import complete")
```

```python
from neo4j import GraphDatabase
import csv

FILE = "cHTS_AllDownloadFiles/cHTS2025_invitrodb42_20250414.txt"

BATCH_SIZE = 5000


driver = GraphDatabase.driver(
    neo4j_uri,
    auth=(neo4j_user, neo4j_password)
)



def insert_batch(tx, rows):

    tx.run("""
    UNWIND $rows AS row
    CREATE (:cHTS_InVitro {
        dataset_name: row.DatasetName,
        record_id: row.RecordID,
        chemical_name: row.ChemicalName,
        casrn: row.CASRN,
        dtxsid: row.DTXSID,
        assay: row.Assay,
        test_range: row.TestRange,
        test_range_unit: row.`TestRange.Unit`,
        endpoint: row.Endpoint,
        response: row.Response,
        response_unit: row.ResponseUnit,
        reference: row.Reference,
        url: row.URL
    })

    """,
    rows=rows
    )


def load():
    with driver.session() as session:
        batch=[]
        with open(
            FILE,
            "r",
            encoding="utf-8"
        ) as f:
            reader = csv.DictReader(
                f,
                delimiter="\t"
            )
            for row in reader:
                batch.append(row)
                if len(batch) == BATCH_SIZE:
                    session.execute_write(
                        insert_batch,
                        batch
                    )
                    print(
                        f"Inserted {len(batch)}"
                    )
                    batch.clear()

            if batch:
                session.execute_write(
                    insert_batch,
                    batch
                )
                print(
                    f"Inserted final {len(batch)}"
                )

load()
driver.close()

print("cHTS import complete")
```

```python
import pandas as pd
from neo4j import GraphDatabase
import csv
import json

with open('neo4j_dbinfo', 'r') as f:
    neo4j_info = json.load(f)

neo4j_uri = neo4j_info["uri"]
neo4j_user = neo4j_info["username"]
neo4j_password = neo4j_info["password"]

df = pd.read_excel(
    "cHTS/cHTSMT_ALL.xlsx",
    sheet_name="Invitrodb42GeneInformation",
    dtype=str
).fillna("")

driver = GraphDatabase.driver(
    neo4j_uri,
    auth=(neo4j_user, neo4j_password)
)

query = """
MERGE (n:cHTSMT {assay: $assay})
SET
    n.assay_source = $assay_source,
    n.species = $species,
    n.tissue = $tissue,
    n.assay_format = $assay_format,
    n.gene = $gene,
    n.entrez_gene_id = $entrez_gene_id
"""

with driver.session() as session:
    for _, row in df.iterrows():

        gene = None if row["Gene"] == "#N/A" else row["Gene"]

        entrez = None
        if row["Entrez Gene ID"] not in ("", "#N/A"):
            entrez = int(row["Entrez Gene ID"])

        session.run(
            query,
            assay=row["Assay"],
            assay_source=row["Assay Source"],
            species=row["Species"],
            tissue=row["Tissue"],
            assay_format=row["Invitro Assay Format"],
            gene=gene,
            entrez_gene_id=entrez,
        )

driver.close()
```

```python
import pandas as pd
from neo4j import GraphDatabase
import json

rows = []

with open('neo4j_dbinfo', 'r') as f:
    neo4j_info = json.load(f)

neo4j_uri = neo4j_info["uri"]
neo4j_user = neo4j_info["username"]
neo4j_password = neo4j_info["password"]

df = pd.read_excel(
    "cHTS/cHTSMT_ALL.xlsx",
    sheet_name="Invitrodb42GeneInformation",
    dtype=str
).fillna("")

driver = GraphDatabase.driver(
    neo4j_uri,
    auth=(neo4j_user, neo4j_password)
)

for _, row in df.iterrows():
    rows.append({
        "assay": row["Assay"],
        "assay_source": row["Assay Source"],
        "species": row["Species"],
        "tissue": row["Tissue"],
        "assay_format": row["Invitro Assay Format"],
        "gene": None if row["Gene"] == "#N/A" else row["Gene"],
        "entrez_gene_id": None if row["Entrez Gene ID"] in ("", "#N/A") else row["Entrez Gene ID"]
    })

query = """
UNWIND $rows AS row
CREATE (n:cHTSMT {
    assay: row.assay,
    assay_source: row.assay_source,
    species: row.species,
    tissue: row.tissue,
    assay_format: row.assay_format,
    gene: row.gene,
    entrez_gene_id: row.entrez_gene_id
})
"""

with driver.session() as session:
    session.run(query, rows=rows)

driver.close()
```
